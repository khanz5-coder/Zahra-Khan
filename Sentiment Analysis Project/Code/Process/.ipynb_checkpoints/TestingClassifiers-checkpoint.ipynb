```python
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df['sentences'])
y = df['code']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 55)
```

```python
model = LogisticRegression(class_weight='balanced', solver="liblinear", C = 0.1, random_state=42)
model.fit(X_train, y_train)
```

```python
ypred = model.predict(X_test)
print("Accuracy_score", accuracy_score(y_test, ypred))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_estimator(model, X_test, y_test)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import confusion_matrix, classification_report
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 55)

# Create a document-term matrix
vectorizer = CountVectorizer()
X1_train = vectorizer.fit_transform(X_train)
X2_test = vectorizer.transform(X_test)
```

```python


nb = MultinomialNB()
nb.fit(X1_train, y_train)
```

```python
classes_pred = nb.predict(X2_test)
```

```python
conf_mat = confusion_matrix(classes_pred, y_test)
conf_mat
```

```python
print("Accuracy_score", accuracy_score(y_test, classes_pred))
```

```python
print(classification_report(y_test, classes_pred))
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.utils.class_weight import compute_sample_weight
sample_weight = compute_sample_weight('balanced', y)

gnb = GaussianNB()
gnb.fit(X1_train.toarray(), y_train)
```

```python
classes_pred = gnb.predict(X2_test.toarray())
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_estimator(gnb, X2_test.toarray(), y_test)
```
