```python
# Importing libraries
#matplotlib inline
import pandas as pd
import numpy as np
import torch
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn import svm
from sklearn.model_selection import GridSearchCV



import re
from textwrap import wrap
from collections import Counter
from string import punctuation
from nltk import word_tokenize, ngrams
from nltk.tokenize import sent_tokenize
```

```python
df = pd.read_csv("/Users/zahrita/Documents/DissData/Output/corpusp1/HandcodOLTUPDATEDcopy.csv", encoding = "latin1")
```

```python
df2 = pd.read_csv("/Users/zahrita/Documents/DissData/Output/corpusp1/HandcodOLT3CA1copy.csv", encoding = "latin1")
```

```python
df3 = pd.read_csv("/Users/zahrita/QTA/QTAhand1 copy.csv", encoding = "latin1")
```

```python
df4 = pd.read_csv("/Users/zahrita/Documents/DissData/Data/Inputs/Handcodev4 copy.csv")
df4 = df4[df4['code']==1]
df5 = pd.read_csv("/Users/zahrita/Documents/DissData/Code/Process/Handcodev5.csv")
df5 = df5[df5['code']==1]
```

```python
df1 = pd.concat([df, df2, df3, df4, df5], ignore_index = True)
df1['code'].value_counts() 
```

```python
df1['sentences'] = df1['sentences'].str.replace(r"[!.\-,]+", "", regex=True)
df1['sentences'] = df1['sentences'].str.replace(r"\s+", " ", regex=True)
```

```python
df1['code'] = df1['code'].replace(5, 1)
df1 = df1.dropna(subset=['code'])
df1['code'] = df1['code'].astype(int)
df1 = df1[df1['code'].isin([0,1])]
print(df1['code'].value_counts())
```

```python
X = df1["sentences"]
y = df1["code"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 52)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression


model = make_pipeline(TfidfVectorizer(
    stop_words="english", min_df = 2, max_df = 0.85, ngram_range=(1,2), max_features = 1500, norm = 'l2'), LogisticRegression(max_iter=1000, class_weight='balanced', solver="liblinear", C = 1.0))
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report
y_pred = model.predict(X_test)
y_tp = model.predict(X_train)
print(classification_report(y_test, y_pred))
test_score = model.score(X_test, y_test)
test_score
```

This has pretty good results for the amount of codes

```python
from sklearn.metrics import roc_curve, auc

fpr, tpr, thresholds = roc_curve(y_test, y_pred)
roc_auc = auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, color='red', lw=2, label='ROC curve (area = %0.2f)' % roc_auc)
plt.plot([0, 1], [0, 1], color='navy',lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver operating characteristic example')
plt.legend(loc='lower right')
plt.show()
```

```python
from sklearn.metrics import precision_recall_curve, auc

precision, recall, thresholds = precision_recall_curve(y_test, y_pred)
pr_auc = auc(precision, recall)
plt.plot(recall, precision, color='red', lw=2, label='PR curve (area = %0.2f)' % pr_auc)
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('PR Curve')
plt.legend(loc='lower left')
plt.show()
```

```python
from sklearn.metrics import matthews_corrcoef
matthews_corrcoef(y_test, y_pred)
```

```python
import matplotlib.pyplot as plt

featname = model.named_steps['tfidfvectorizer'].get_feature_names_out()
coefs = model.named_steps['logisticregression'].coef_[0]
top_nativst = sorted(zip(coefs, featname))[-20:]

coefvals, featnames = zip(*top_nativst)
colors = ['#186FA6' if v >= 0 else '#993C1D' for v in coefvals]

fig, ax = plt.subplots(figsize=(8, 6))
ax.barh(featnames, coefvals, color = colors)
ax.axvline(0, color = 'gray', linewidth=0.8, linestyle='--')
ax.set_xlabel('Coefficient Values')
ax.set_title("Top 15 features by logit regression coefficients")
plt.tight_layout()
plt.show()
```

## Spotchecks:
In order to determine the capability of the model and determine if applying sentence handcodes to articles is sufficent, or if I reiqure a sentence tokenizers, I ran spot checks with both methods and found they report the same, and the spot checks on the model siply applied to articles has a confidence range of about 51-80% when classifying class 1.

```python
dfcorp = pd.read_csv("~/Documents/DissData/Data/Inputs/article_olv7.csv")
dfcorp.head()
```

```python
sample = dfcorp.sample(250, random_state=40)
sample['pred'] = model.predict(sample['articles'])
probas = model.predict_proba(sample['articles'])
sample['confint'] = probas.max(axis=1)

sample[['articles', 'pred', 'confint']].to_csv('spot_checks8.csv', index=False)
```

```python
def classify_articles(articles):
    sentences = sent_tokenize(articles)
    probs = model.predict_proba(sentences)
    averageprob = probs.mean(axis=0)
    return model.classes_[averageprob.argmax()]
    
sample['predicted_class'] = sample["articles"].apply(classify_articles)
print(sample['predicted_class'].value_counts())
print(sample)
```
